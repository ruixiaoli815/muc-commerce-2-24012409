## 任务5：描述性统计与一级品类汇总

```python
# TODO 1：使用describe查看商品价格摘要，变量名为price_summary
price_summary = None

# TODO 2：按一级品类统计商品数、平均价格和中位价格
# 按平均价格从高到低排序，变量名为category_summary
category_summary = None

# TODO 3：展示结果
# TODO 1: 使用describe查看商品价格摘要，变量名为price_summary
price_summary = df['商品价格'].describe().round(2)

# TODO 2: 按一级品类统计商品数、平均价格和中位价格
# 按平均价格从高到低排序，变量名为category_summary
category_summary = (
    df.groupby('一级品类')
      .agg(商品数=('商品id', 'size'),
           平均价格=('商品价格', 'mean'),
           中位价格=('商品价格', 'median'))
      .sort_values('平均价格', ascending=False)
      .round(2)
)

# TODO 3: 展示结果
prices = df['商品价格'].to_numpy()
print('\n价格均值：', round(prices.mean(), 2))
print('价格中位数：', round(np.median(prices), 2))
print('价格不低于 1000 元的商品数：', int((prices >= 1000).sum()))
print('\n商品价格描述统计：')
print(price_summary)
print('\n一级品类商品数：')
print(df['一级品类'].value_counts())
print('\n一级品类汇总：')
print(category_summary)
```

```python
# 检查点5
assert isinstance(price_summary, pd.Series)
assert isinstance(category_summary, pd.DataFrame)
assert {"商品数", "平均价格", "中位价格"}.issubset(category_summary.columns)
assert category_summary["商品数"].sum() == len(df)
assert category_summary["平均价格"].is_monotonic_decreasing
assert abs(df["商品价格"].mean() - 938.26) < 0.02
print("检查点5通过")
```

## 挑战任务：省份—类别分析

```python
# TODO 1：选择两个省份
provinces = ["广东", "江苏"]

# TODO 2：筛选省份并统计商品数、平均价格和中位价格
province_summary = None

# TODO 3：分别计算两个省份最常见的一级品类
top_categories = None

# TODO 4：展示结果
# TODO 1: 选择两个省份
provinces = ['广东', '江苏']

# TODO 2: 筛选省份并统计商品数、平均价格和中位价格
subset = df[df['省份'].isin(provinces)]
province_summary = (
    subset.groupby('省份')
          .agg(商品数=('商品id', 'size'),
               平均价格=('商品价格', 'mean'),
               中位价格=('商品价格', 'median'))
          .round(2)
)

# TODO 3: 分别计算两个省份最常见的一级品类
rows = []
for p in provinces:
    top = subset.loc[subset['省份'] == p, '一级品类'].value_counts().head(1)
    rows.append({'省份': p, '最常见一级品类': top.index[0], '该类商品数': int(top.iloc[0])})
top_categories = pd.DataFrame(rows)

# TODO 4: 展示结果
print('\n省份对比汇总：')
print(province_summary)
print('\n各省份最常见一级品类：')
print(top_categories)
category_summary.to_csv('上午_一级品类价格汇总.csv', encoding='utf-8-sig')
province_summary.to_csv('挑战_省份对比汇总.csv', encoding='utf-8-sig')
```

```python
# 检查点6
assert len(provinces) == 2 and provinces[0] != provinces[1]
assert isinstance(province_summary, pd.DataFrame)
assert set(province_summary.index) == set(provinces)
assert {"商品数", "平均价格", "中位价格"}.issubset(province_summary.columns)
assert isinstance(top_categories, pd.DataFrame)
print("检查点6通过")
```
